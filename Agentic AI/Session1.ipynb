```python
## Task 1: AI Agent vs Traditional Machine Learning Model

A traditional machine learning model usually makes a prediction based on given input, while an AI agent can reason, make decisions, use tools, and take multiple actions to achieve a goal. For example, an Instagram recommendation model may predict which post a user will like, while an AI agent could decide what content to show, search for relevant content, and adjust recommendations based on the user's ongoing behavior.
```

```python
## Task 2: Characteristics of Agent-Based AI Systems

| Characteristic | Example in an App |
|---|---|
| **Autonomy** | In Flipkart, an AI agent could automatically track an order and notify the customer about delays without the user asking. |
| **Decision Making** | In Swiggy, an AI agent could compare restaurants, delivery time, price, and user preferences and select the best option. |
| **Tool Usage & Action** | In BookMyShow, an AI agent could search for suitable movies, check available seats, select the best seats, and proceed with booking based on the user's preferences. |

These characteristics allow AI agents to go beyond simply providing predictions or answers and actively work toward completing a user's goal.
```

```python
## Task 3: ReAct Agents

ReAct (Reasoning + Acting) is an approach where an AI agent combines reasoning with actions. Instead of only generating an answer, the agent can decide what action to take, use an external tool or source, observe the result, and then decide what to do next. This creates an iterative loop of reasoning, acting, observing, and updating the plan.

AutoGPT follows a similar agentic idea by allowing an AI system to break a larger goal into smaller tasks, execute actions using available tools, use the results to guide the next step, and continue until the goal is completed.

### Example: AI Agent in Spotify

A Spotify AI agent could work as a personal music assistant. If a user usually listens to workout music at 7 AM, the agent could proactively analyze the user's listening history, current preferences, and available songs, create a suitable playlist, and automatically update it over time without requiring the user to manually search for songs.

The key difference is that the agent does not only recommend one song based on an input; it can make decisions, use information, take actions, observe results, and continue working toward the user's goal.
```

```python
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch

fig, ax = plt.subplots(figsize=(12, 8))
ax.set_xlim(0, 10)
ax.set_ylim(0, 12)
ax.axis("off")

def box(x, y, w, h, text):
    patch = FancyBboxPatch(
        (x, y), w, h,
        boxstyle="round,pad=0.08",
        linewidth=1.5
    )
    ax.add_patch(patch)
    ax.text(
        x + w/2, y + h/2, text,
        ha="center", va="center",
        fontsize=11, wrap=True
    )

def arrow(x1, y1, x2, y2):
    ax.add_patch(
        FancyArrowPatch(
            (x1, y1), (x2, y2),
            arrowstyle="->",
            mutation_scale=18,
            linewidth=1.5
        )
    )

box(3.5, 10, 3, 1, "User gives a goal\n'Plan my day'")
box(3.5, 8, 3, 1, "ChatGPT Agent\nUnderstands the goal")
box(3.5, 6, 3, 1, "Reason & Plan\nBreaks goal into tasks")
box(1, 4, 3, 1, "Use Tools\nCalendar / Search / Email")
box(6, 4, 3, 1, "Take Actions\nCreate / Update / Send")
box(3.5, 2, 3, 1, "Observe Results\nCheck what happened")
box(3.5, 0, 3, 1, "Proactive Follow-up\nContinue or notify user")

arrow(5, 10, 5, 9)
arrow(5, 8, 5, 7)
arrow(5, 6, 2.5, 5)
arrow(5, 6, 7.5, 5)
arrow(2.5, 4, 5, 3)
arrow(7.5, 4, 5, 3)
arrow(5, 2, 5, 1)

ax.text(
    5, 11.3,
    "ChatGPT Agent – Autonomous & Proactive Workflow",
    ha="center", fontsize=16, fontweight="bold"
)

plt.show()
```
