```python
def fetch_menu():
    menu = [
        {"dish": "Pizza", "price": 250},
        {"dish": "Burger", "price": 150},
        {"dish": "Pasta", "price": 200}
    ]
    print("Agent 1: Menu fetched")
    return menu


def analyze_orders(menu):
    total_items = len(menu)
    total_value = sum(item["price"] for item in menu)

    analysis = {
        "total_items": total_items,
        "total_value": total_value
    }

    print("Agent 2: Orders analyzed")
    return analysis


def generate_delivery_report(analysis):
    report = (
        f"Total Items: {analysis['total_items']}\n"
        f"Total Order Value: ₹{analysis['total_value']}\n"
        f"Delivery Report: Ready"
    )

    print("Agent 3: Report generated")
    return report


menu = fetch_menu()
analysis = analyze_orders(menu)
report = generate_delivery_report(analysis)

print("\nFinal Delivery Report")
print(report)
```

```python
def agent_1_fetch_scores():
    return {
        "Mumbai Indians": 198,
        "Chennai Super Kings": 185,
        "Royal Challengers Bengaluru": 210,
        "Kolkata Knight Riders": 176
    }


def agent_2_find_highest_score(scores):
    team = max(scores, key=scores.get)
    return {
        "team": team,
        "score": scores[team]
    }


def agent_3_generate_report(result):
    return (
        "IPL Match Score Report\n"
        "----------------------\n"
        f"Highest Scoring Team: {result['team']}\n"
        f"Highest Score: {result['score']}"
    )


scores = agent_1_fetch_scores()
highest_score = agent_2_find_highest_score(scores)
report = agent_3_generate_report(highest_score)

print(report)
```
