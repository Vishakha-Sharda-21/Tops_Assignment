```python
def call_weather_api(city_name):
    weather_data = {
        "Ahmedabad": 32,
        "Mumbai": 29,
        "Delhi": 35,
        "Bengaluru": 24
    }

    temperature = weather_data.get(city_name.title(), 28)

    return {
        "city": city_name.title(),
        "temperature": temperature,
        "unit": "Celsius"
    }


def weather_agent(city):
    result = call_weather_api(city)
    return f"The temperature in {result['city']} is {result['temperature']}°C."


city = input("Enter city: ")
print(weather_agent(city))
```

```python
def recommend_restaurant(cuisine):
    print(f"1. Finding the best {cuisine} restaurants...")
    return "Best Rated Restaurant"


def place_order(restaurant, dish):
    print(f"2. Ordering {dish} from {restaurant}...")
    return "Order ID: ZOM12345"


def confirm_delivery(order_id):
    print(f"3. Confirming delivery for {order_id}...")
    return "Delivery confirmed"


cuisine = input("Enter preferred cuisine: ")
dish = input("Enter desired dish: ")

restaurant = recommend_restaurant(cuisine)
order_id = place_order(restaurant, dish)
confirmation = confirm_delivery(order_id)

print(confirmation)
```

```python
def decide_playlist_action(song_likes, song_skips):
    if song_likes >= 100 and song_skips <= 10:
        return "promote"
    elif song_skips >= 50 and song_likes < 20:
        return "remove"
    else:
        return "recommend"


print(decide_playlist_action(150, 5))
print(decide_playlist_action(10, 60))
print(decide_playlist_action(50, 20))
```

```python
from PyPDF2 import PdfReader
import re


def read_pdf(file_path):
    reader = PdfReader(file_path)
    text = ""

    for page in reader.pages:
        text += page.extract_text() or ""

    return text


def find_top_scorer(text):
    pattern = r"([A-Za-z .]+)\s*[-:]\s*(\d+)\s*runs"
    matches = re.findall(pattern, text, re.IGNORECASE)

    if not matches:
        return "Top scorer could not be identified."

    player, runs = max(matches, key=lambda x: int(x[1]))
    return f"{player.strip()} scored the most runs with {runs} runs."


def ipl_agent(file_path, question):
    text = read_pdf(file_path)

    if "most runs" in question.lower():
        return find_top_scorer(text)

    return "I can currently answer questions about the highest run scorer."


pdf_file = "ipl_match_stats.pdf"
question = input("Ask a question: ")

print(ipl_agent(pdf_file, question))
```

```python
from PyPDF2 import PdfReader
import re


def extract_ticket_info(pdf_path):
    reader = PdfReader(pdf_path)
    text = ""

    for page in reader.pages:
        text += page.extract_text() or ""

    movie = re.search(
        r"Movie\s*[:\-]\s*(.+)",
        text,
        re.IGNORECASE
    )

    seat = re.search(
        r"Seat(?:\s*Number)?\s*[:\-]\s*([A-Z0-9,\s]+)",
        text,
        re.IGNORECASE
    )

    showtime = re.search(
        r"Show\s*Time\s*[:\-]\s*(.+)",
        text,
        re.IGNORECASE
    )

    return {
        "movie_name": movie.group(1).strip() if movie else "Not found",
        "seat_number": seat.group(1).strip() if seat else "Not found",
        "showtime": showtime.group(1).strip() if showtime else "Not found"
    }


ticket = extract_ticket_info("bookmyshow_ticket.pdf")

print(ticket)
```
