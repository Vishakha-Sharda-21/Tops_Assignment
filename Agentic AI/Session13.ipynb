```python
from pathlib import Path

def multimodal_agent(text_prompt, image_path):
    image_file = Path(image_path)

    print("Text Prompt:", text_prompt)
    print("Image Filename:", image_file.name)
    print("Both text and image were received successfully.")


text_prompt = input("Enter your text prompt: ")
image_path = input("Enter image file path: ")

multimodal_agent(text_prompt, image_path)
```

```python
import os
from dotenv import load_dotenv
import requests

def fetch_song_info(song_name):
    load_dotenv()
    access_token = os.getenv("SPOTIFY_ACCESS_TOKEN", "YOUR_SPOTIFY_ACCESS_TOKEN")

    url = "https://api.spotify.com/v1/search"

    headers = {
        "Authorization": f"Bearer {access_token}"
    }

    params = {
        "q": song_name,
        "type": "track",
        "limit": 1
    }

    response = requests.get(url, headers=headers, params=params)

    if response.status_code != 200:
        print("API Error:", response.status_code)
        return

    data = response.json()
    tracks = data.get("tracks", {}).get("items", [])

    if not tracks:
        print("Song not found.")
        return

    track = tracks[0]

    print("Song:", track["name"])
    print("Artist:", track["artists"][0]["name"])
    print("Album:", track["album"]["name"])


song = input("Enter song name: ")
fetch_song_info(song)
```
