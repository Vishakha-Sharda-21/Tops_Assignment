````python
from flask import Flask, request, jsonify

app = Flask(__name__)

orders = {
    "ORD1001": "Delivered",
    "ORD1002": "In Transit",
    "ORD1003": "Cancelled"
}

users = {
    "U001": {
        "name": "Rahul Sharma",
        "email": "rahul@example.com",
        "phone": "9876543210"
    },
    "U002": {
        "name": "Priya Patel",
        "email": "priya@example.com",
        "phone": "9876543211"
    }
}


@app.route("/")
def home():
    return "MCP Server Running"


@app.route("/order-status", methods=["POST"])
def order_status():
    data = request.get_json(silent=True) or {}

    if "orderId" not in data:
        return jsonify({
            "error": "orderId is required"
        }), 400

    order_id = data["orderId"]
    status = orders.get(order_id, "Order Not Found")

    return jsonify({
        "orderId": order_id,
        "status": status
    })


@app.route("/notify", methods=["POST"])
def notify():
    data = request.get_json(silent=True) or {}

    if "userId" not in data or "message" not in data:
        return jsonify({
            "error": "userId and message are required"
        }), 400

    user_id = data["userId"]
    message = data["message"]

    with open("notifications.log", "a", encoding="utf-8") as file:
        file.write(f"User: {user_id} | Message: {message}\n")

    return jsonify({
        "message": "Notification logged successfully"
    })


@app.route("/user-profile", methods=["GET"])
def user_profile():
    user_id = request.args.get("userId")

    if not user_id:
        return jsonify({
            "error": "userId is required"
        }), 400

    profile = users.get(user_id)

    if not profile:
        return jsonify({
            "error": "User not found"
        }), 404

    return jsonify({
        "userId": user_id,
        **profile
    })


if __name__ == "__main__":
    app.run(host="127.0.0.1", port=5000, debug=False)
````
