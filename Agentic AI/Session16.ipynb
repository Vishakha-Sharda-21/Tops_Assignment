````python
from langchain_openai import ChatOpenAI

teams = {
    "Mumbai Indians": "Hardik Pandya",
    "Chennai Super Kings": "Ruturaj Gaikwad",
    "Royal Challengers Bengaluru": "Rajat Patidar",
    "Kolkata Knight Riders": "Ajinkya Rahane",
    "Rajasthan Royals": "Sanju Samson",
    "Sunrisers Hyderabad": "Pat Cummins"
}

llm = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0
)

def ipl_agent(question):
    team_info = "\n".join(
        f"{team}: {captain}"
        for team, captain in teams.items()
    )

    prompt = f"""
Answer the user's question using only this IPL team information:

{team_info}

User question: {question}

If the team is not in the list, say that the information is not available.
"""

    response = llm.invoke(prompt)
    return response.content


question = input("Ask about an IPL team: ")
print(ipl_agent(question))
````

````python
from typing import TypedDict
from langgraph.graph import StateGraph, START, END

restaurants = {
    "ahmedabad": [
        "The Green House",
        "Agashiye",
        "Swati Snacks"
    ],
    "mumbai": [
        "Trishna",
        "Britannia & Co.",
        "Leopold Cafe"
    ],
    "delhi": [
        "Indian Accent",
        "Bukhara",
        "Karim's"
    ]
}

class RestaurantState(TypedDict):
    city: str
    recommendations: list


def get_city(state):
    city = input("Enter your city: ")
    return {"city": city.lower()}


def recommend_restaurants(state):
    city = state["city"]
    return {
        "recommendations": restaurants.get(
            city,
            ["No restaurants found for this city."]
        )
    }


def show_recommendations(state):
    print(f"\nRestaurants in {state['city'].title()}:")
    for restaurant in state["recommendations"]:
        print("-", restaurant)


builder = StateGraph(RestaurantState)

builder.add_node("get_city", get_city)
builder.add_node("recommend", recommend_restaurants)
builder.add_node("show", show_recommendations)

builder.add_edge(START, "get_city")
builder.add_edge("get_city", "recommend")
builder.add_edge("recommend", "show")
builder.add_edge("show", END)

graph = builder.compile()

graph.invoke({
    "city": "",
    "recommendations": []
})
````

````python
from typing import TypedDict
from langchain_openai import ChatOpenAI
from langgraph.graph import StateGraph, START, END

llm = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0
)

class MovieState(TypedDict):
    review: str
    summary: str
    wants_details: str
    response: str


def summarize_review(state):
    prompt = f"""
    Summarize this BookMyShow movie review in 2 sentences:

    {state['review']}
    """

    result = llm.invoke(prompt)

    return {
        "summary": result.content
    }


def ask_user(state):
    print("\nMovie Summary:")
    print(state["summary"])

    answer = input("\nDo you want more details? (yes/no): ")

    return {
        "wants_details": answer.lower()
    }


def follow_up(state):
    if state["wants_details"] == "yes":
        response = "Here are more details about the movie, including its cast, genre, rating, and story."
    else:
        response = "Okay. No additional details requested."

    return {
        "response": response
    }


def show_response(state):
    print("\nAgent:", state["response"])


builder = StateGraph(MovieState)

builder.add_node("summarize", summarize_review)
builder.add_node("ask_user", ask_user)
builder.add_node("follow_up", follow_up)
builder.add_node("show", show_response)

builder.add_edge(START, "summarize")
builder.add_edge("summarize", "ask_user")
builder.add_edge("ask_user", "follow_up")
builder.add_edge("follow_up", "show")
builder.add_edge("show", END)

graph = builder.compile()

review = """
The movie has an engaging story and excellent performances by the lead actors.
The cinematography is impressive, although the second half feels slightly slow.
Overall, it is an enjoyable movie worth watching.
"""

graph.invoke({
    "review": review,
    "summary": "",
    "wants_details": "",
    "response": ""
})
````

````python
## Changes Made

1. Added a separate `get_top_trends()` function to organize the workflow.
2. Kept the Twitter API as a mock function so no external API is required.
3. Changed the prompt to explicitly request exactly 2 numbered trends.
4. Set the temperature to 0 to make the output more consistent.
5. The final code prints only the top 2 trends in the console.
````
