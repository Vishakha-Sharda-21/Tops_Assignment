````python
class SearchAgent:
    def __init__(self):
        self.products = [
            "iPhone 15",
            "Samsung Galaxy S24",
            "OnePlus 12",
            "Sony Headphones",
            "Dell Laptop",
            "HP Laptop",
            "Apple Watch",
            "Boat Earbuds"
        ]

    def search(self, query):
        return [
            product for product in self.products
            if query.lower() in product.lower()
        ]


class RecommendationAgent:
    def recommend(self, user_id):
        recommendations = {
            "U001": ["iPhone 15", "Apple Watch", "Sony Headphones"],
            "U002": ["Samsung Galaxy S24", "Boat Earbuds", "Dell Laptop"],
            "U003": ["OnePlus 12", "HP Laptop", "Apple Watch"]
        }

        return recommendations.get(
            user_id,
            ["OnePlus 12", "Sony Headphones", "Apple Watch"]
        )


search_agent = SearchAgent()
recommendation_agent = RecommendationAgent()

print("Search Results:")
print(search_agent.search("Laptop"))

print("\nRecommendations:")
print(recommendation_agent.recommend("U001"))
````

````python
class ChatAgent:
    def handle_query(self, query, manager):
        if "top trending playlist" in query.lower():
            manager.log.append("ChatAgent → PlaylistAgent: Get trending playlist")
            
            trending = manager.playlist_agent.get_trending()

            manager.log.append(
                f"PlaylistAgent → ChatAgent: {trending}"
            )

            return f"Top trending playlist: {trending}"

        if "playlist" in query.lower():
            manager.log.append("ChatAgent → PlaylistAgent: Create playlist")

            playlist = manager.playlist_agent.create_playlist(query)

            manager.log.append(
                f"PlaylistAgent → ChatAgent: {playlist}"
            )

            return f"Playlist: {playlist}"

        return "I can help with playlists."


class PlaylistAgent:
    def create_playlist(self, request):
        if "workout" in request.lower():
            return ["Believer", "Thunder", "Till I Collapse"]

        if "chill" in request.lower():
            return ["Perfect", "Photograph", "Until I Found You"]

        return ["Blinding Lights", "Levitating", "Shape of You"]

    def get_trending(self):
        return [
            "APT.",
            "Die With A Smile",
            "Espresso"
        ]


class MultiAgentManager:
    def __init__(self):
        self.chat_agent = ChatAgent()
        self.playlist_agent = PlaylistAgent()
        self.log = []

    def process(self, query):
        return self.chat_agent.handle_query(query, self)


manager = MultiAgentManager()

query = input("You: ")

print("\nAssistant:")
print(manager.process(query))

print("\nAgent Communication Log:")
for message in manager.log:
    print(message)
````
