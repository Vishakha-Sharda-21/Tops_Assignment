```python
def perceive(last_song):
    return last_song


def think(last_song):
    genres = {
        "Blinding Lights": "Pop",
        "Believer": "Rock",
        "Shape of You": "Pop",
        "Perfect": "Romantic"
    }
    return genres.get(last_song, "Pop")


def act(genre):
    recommendations = {
        "Pop": "Espresso",
        "Rock": "Thunder",
        "Romantic": "Photograph"
    }
    return recommendations.get(genre, "APT.")


def learn(preferences, genre):
    preferences[genre] = preferences.get(genre, 0) + 1
    return preferences


preferences = {}

last_song = perceive("Blinding Lights")
next_genre = think(last_song)
recommended_song = act(next_genre)
preferences = learn(preferences, next_genre)

print("Last played song:", last_song)
print("Next genre:", next_genre)
print("Suggested song:", recommended_song)
print("Updated preferences:", preferences)
```

```python
Perceive → Understand the user's message, order details, and current problem.

Think → Analyze the request and decide the appropriate solution or next action.

Act → Provide a response or perform an action such as checking order status or issuing a refund.

Learn → Use the conversation outcome and feedback to improve future support responses.
```

```python
class ShortTermMemory:
    def __init__(self, limit=5):
        self.limit = limit
        self.messages = []

    def add_message(self, message):
        self.messages.append(message)

        if len(self.messages) > self.limit:
            self.messages.pop(0)

    def get_memory(self):
        return self.messages


memory = ShortTermMemory()

messages = [
    "Hello",
    "Show me Indian restaurants",
    "Under 500 rupees",
    "Near me",
    "I want vegetarian food",
    "Also show highly rated restaurants"
]

for message in messages:
    memory.add_message(message)

print("Current Memory:")
for message in memory.get_memory():
    print("-", message)
```

```python
long_term_memory = {}


def add_preference(user_id, key, value):
    if user_id not in long_term_memory:
        long_term_memory[user_id] = {}

    long_term_memory[user_id][key] = value


def get_preferences(user_id):
    return long_term_memory.get(user_id, {})


add_preference("U001", "favorite_cuisine", "Indian")
add_preference("U001", "payment_method", "UPI")
add_preference("U002", "favorite_cuisine", "Italian")

print("U001:", get_preferences("U001"))
print("U002:", get_preferences("U002"))
```
