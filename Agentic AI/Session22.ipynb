```python
import requests

def search_restaurants(query):
    print("User Input:", query)

    # Mock API endpoint for demonstration
    url = "https://example.com/api/restaurants"

    print("API Call:", url)

    try:
        response = requests.get(url, params={"query": query}, timeout=5)
        response.raise_for_status()
        data = response.json()

        restaurants = data.get("restaurants", [])[:3]

        print("Top 3 Restaurants:")
        for restaurant in restaurants:
            print("-", restaurant["name"])

    except requests.RequestException:
        print("API unavailable. Using mock results.")
        restaurants = [
            "La Pino's Pizza",
            "Domino's Pizza",
            "Pizza Hut"
        ]

        print("Top 3 Restaurants:")
        for restaurant in restaurants:
            print("-", restaurant)


query = input("Enter your restaurant search: ")
search_restaurants(query)
```

```python
from collections import Counter

def playlist_agent(filename):
    with open(filename, "r", encoding="utf-8") as file:
        songs = file.readlines()

    artists = []

    for song in songs:
        if " - " in song:
            artist = song.strip().split(" - ")[-1]
            artists.append(artist)

    most_common_artist = Counter(artists).most_common(1)[0][0]

    recommendations = {
        "The Weeknd": "Die For You",
        "Imagine Dragons": "Radioactive",
        "Ed Sheeran": "Perfect"
    }

    suggestion = recommendations.get(
        most_common_artist,
        "A new trending song"
    )

    print("Most common artist:", most_common_artist)
    print("Suggested song:", suggestion)


playlist_agent("my_playlist.txt")
```

```python
def movie_recommendation():
    genre = input("What movie genre do you prefer? ")

    question = "Do you prefer Hollywood or Bollywood?"
    print(question)

    industry = input("Your choice: ")

    recommendations = {
        ("action", "hollywood"): "John Wick",
        ("action", "bollywood"): "War",
        ("comedy", "hollywood"): "The Hangover",
        ("comedy", "bollywood"): "Hera Pheri",
        ("sci-fi", "hollywood"): "Interstellar",
        ("sci-fi", "bollywood"): "Ra.One"
    }

    key = (genre.lower(), industry.lower())

    movie = recommendations.get(
        key,
        "3 Idiots"
    )

    print("Recommended movie:", movie)


movie_recommendation()
```

```python
memory = {
    "last_search": None,
    "last_result": None
}


def search_cricket_scores(query):
    scores = {
        "ipl": "Mumbai Indians 185/5 vs Chennai Super Kings 172/8",
        "rcb": "RCB 190/4 vs Punjab Kings 175/7"
    }

    return scores.get(
        query.lower(),
        "No score found"
    )


def run_agent():
    query = input("Search cricket scores: ")

    result = search_cricket_scores(query)

    memory["last_search"] = query
    memory["last_result"] = result

    print("\nResult:", result)
    print("Updated Memory:", memory)


run_agent()
```
