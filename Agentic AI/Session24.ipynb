```python
import requests

def get_weather(city, latitude, longitude):
    url = "https://api.open-meteo.com/v1/forecast"

    params = {
        "latitude": latitude,
        "longitude": longitude,
        "current": "temperature_2m"
    }

    response = requests.get(url, params=params, timeout=10)
    response.raise_for_status()

    data = response.json()

    temperature = data["current"]["temperature_2m"]
    unit = data["current_units"]["temperature_2m"]

    print(f"City: {city}")
    print(f"Current Temperature: {temperature}{unit}")


get_weather("Ahmedabad", 23.0225, 72.5714)
```

```python
import csv

def calculate_food_expenses(filename):
    total = 0

    with open(filename, "r", encoding="utf-8") as file:
        reader = csv.DictReader(file)

        for row in reader:
            if row["category"].strip().lower() == "food":
                total += float(row["amount"])

    print(f"Total Food Expenses: ₹{total:.2f}")


calculate_food_expenses("my_expenses.csv")
```

```python
import requests
import os
from dotenv import load_dotenv

def get_technology_news(api_key):
    url = "https://gnews.io/api/v4/top-headlines"

    params = {
        "category": "technology",
        "lang": "en",
        "country": "us",
        "max": 3,
        "apikey": api_key
    }

    response = requests.get(url, params=params, timeout=10)
    response.raise_for_status()

    articles = response.json().get("articles", [])

    print("Top 3 Technology News:\n")

    for i, article in enumerate(articles[:3], 1):
        print(f"{i}. {article['title']}")


load_dotenv()
get_technology_news(os.getenv("GNEWS_API_KEY", "YOUR_GNEWS_API_KEY"))
```

```python
import re

def calculate(expression):
    expression = expression.replace(" ", "")

    if not re.fullmatch(r"\d+(?:\.\d+)?(?:[+\-*/]\d+(?:\.\d+)?)*", expression):
        return "Invalid expression"

    numbers = re.split(r"[+\-*/]", expression)
    operators = re.findall(r"[+\-*/]", expression)

    result = float(numbers[0])

    for operator, number in zip(operators, numbers[1:]):
        number = float(number)

        if operator == "+":
            result += number
        elif operator == "-":
            result -= number
        elif operator == "*":
            result *= number
        elif operator == "/":
            if number == 0:
                return "Cannot divide by zero"
            result /= number

    return int(result) if result.is_integer() else result


expression = input("Enter expression: ")

print("Result:", calculate(expression))
```
