```python
class FoodOrderAgent:
    def __init__(self):
        self.order = None
        self.status = "No active order"

    def place_order(self, items):
        self.order = items
        self.status = "Order Placed"
        print(f"Order placed: {items}")

    def check_status(self):
        print(f"Order status: {self.status}")

    def cancel_order(self):
        if self.order:
            self.status = "Order Cancelled"
            print("Order cancelled successfully.")
        else:
            print("No active order.")


agent = FoodOrderAgent()

agent.place_order(["Pizza", "Garlic Bread", "Coke"])
agent.check_status()
agent.cancel_order()
agent.check_status()
```

```python
class MovieRecommendationAgent:
    def recommend(self, genre, rating):
        genre = genre.lower()

        if genre == "action":
            if rating >= 8:
                movie = "Inception"
            else:
                movie = "John Wick"

        elif genre == "comedy":
            if rating >= 8:
                movie = "The Grand Budapest Hotel"
            else:
                movie = "Free Guy"

        elif genre == "sci-fi":
            if rating >= 8:
                movie = "Interstellar"
            else:
                movie = "Ready Player One"

        else:
            movie = "The Shawshank Redemption"

        print(f"Recommended movie: {movie}")


agent = MovieRecommendationAgent()

genre = input("Enter preferred genre: ")
rating = float(input("Enter minimum rating: "))

agent.recommend(genre, rating)
```

```python
class SongAgent:
    def get_song(self, mood):
        songs = {
            "happy": "Happy - Pharrell Williams",
            "sad": "Someone Like You - Adele",
            "energetic": "Believer - Imagine Dragons",
            "relaxed": "Perfect - Ed Sheeran"
        }

        return songs.get(mood.lower(), "No song found")


class PlaylistAgent:
    def __init__(self, song_agent):
        self.song_agent = song_agent

    def request_song(self, mood):
        print(f"PlaylistAgent → SongAgent: Find a {mood} song")
        song = self.song_agent.get_song(mood)
        print(f"SongAgent → PlaylistAgent: {song}")
        return song


song_agent = SongAgent()
playlist_agent = PlaylistAgent(song_agent)

playlist_agent.request_song("happy")
```

```python
class ShoppingAgent:

    def check_preferences(self, user, item):
        return item["category"] in user["preferences"]

    def check_availability(self, item):
        return item["stock"] > 0

    def decide_action(self, user, item):
        if not self.check_preferences(user, item):
            return "Item does not match user preferences"

        if not self.check_availability(item):
            return "Item is out of stock"

        return "Add item to cart"


agent = ShoppingAgent()

user = {
    "preferences": ["electronics", "books"]
}

item = {
    "name": "Wireless Headphones",
    "category": "electronics",
    "stock": 5
}

print(agent.decide_action(user, item))
```

```python
class ShoppingCartAgent:
    def __init__(self):
        self.cart = {}

    def add_item(self, name, price, quantity=1):
        self.cart[name] = {
            "price": price,
            "quantity": quantity
        }

    def remove_item(self, name):
        if name in self.cart:
            del self.cart[name]

    def get_total(self):
        return sum(
            item["price"] * item["quantity"]
            for item in self.cart.values()
        )
```
