```python
class MusicAgent:
    def __init__(self):
        self.trending = {
            "Pop": ["APT.", "Espresso", "Die With A Smile"],
            "Rock": ["Believer", "Thunder", "Radioactive"],
            "Hip-Hop": ["God's Plan", "HUMBLE.", "Lose Yourself"]
        }

    def perceive(self, history):
        genres = [song["genre"] for song in history]
        return list(set(genres))

    def reason(self, preferred_genres):
        recommendations = []

        for genre in preferred_genres:
            recommendations.extend(self.trending.get(genre, []))

        return recommendations[:3]

    def act(self, recommendations):
        print("Recommended Playlist:")
        for song in recommendations:
            print("-", song)


agent = MusicAgent()

history = [
    {"song": "Levitating", "genre": "Pop"},
    {"song": "Blinding Lights", "genre": "Pop"},
    {"song": "Believer", "genre": "Rock"}
]

genres = agent.perceive(history)
recommendations = agent.reason(genres)
agent.act(recommendations)
```

```python
class MovieReleaseAgent:
    def fetch_movies(self):
        return [
            {"title": "Avatar 3", "genre": "Sci-Fi"},
            {"title": "Deadpool 3", "genre": "Action"},
            {"title": "Inside Out 2", "genre": "Animation"},
            {"title": "A Quiet Place", "genre": "Horror"},
            {"title": "The Batman", "genre": "Action"}
        ]


class MovieRecommendationAgent:
    def recommend(self, movies, preferred_genres):
        return [
            movie for movie in movies
            if movie["genre"].lower() in
            [genre.lower() for genre in preferred_genres]
        ]


release_agent = MovieReleaseAgent()
recommendation_agent = MovieRecommendationAgent()

movies = release_agent.fetch_movies()

print("MovieReleaseAgent → MovieRecommendationAgent")
print("Sending new movie releases...\n")

recommendations = recommendation_agent.recommend(
    movies,
    ["Action", "Sci-Fi"]
)

print("Recommended Movies:")
for movie in recommendations:
    print("-", movie["title"])
```

```python
class AgentRestaurantRecommender:

    def perceive(self, user_input):
        return user_input.strip().lower()

    def reason(self, cuisine):
        restaurants = {
            "italian": ["La Piazza", "Little Italy", "Olive Garden"],
            "indian": ["Agashiye", "Punjab Grill", "Barbeque Nation"],
            "chinese": ["Mainland China", "Wow China", "China Garden"],
            "mexican": ["Mexican Bowl", "Taco Bell", "Chili's"]
        }

        return restaurants.get(cuisine, [])

    def act(self, restaurants):
        if restaurants:
            print("Recommended Restaurants:")
            for restaurant in restaurants:
                print("-", restaurant)
        else:
            print("No restaurants found for this cuisine.")


agent = AgentRestaurantRecommender()

user_input = input("Enter cuisine: ")

cuisine = agent.perceive(user_input)
restaurants = agent.reason(cuisine)
agent.act(restaurants)
```
