```python
def restaurant_agent():
    preference = input("What type of food do you prefer? ")

    if "pizza" in preference.lower() or "italian" in preference.lower():
        restaurant = "Italian Restaurant"
    elif "indian" in preference.lower() or "spicy" in preference.lower():
        restaurant = "Indian Restaurant"
    elif "chinese" in preference.lower():
        restaurant = "Chinese Restaurant"
    elif "healthy" in preference.lower():
        restaurant = "Healthy Food Restaurant"
    else:
        restaurant = "Multi-Cuisine Restaurant"

    print(f"Suggested restaurant type: {restaurant}")

    feedback = input("Was this suggestion useful? (yes/no): ")

    if feedback.lower() == "yes":
        print("Learning: I will continue recommending similar restaurants.")
    else:
        print("Learning: I will try a different restaurant type next time.")

restaurant_agent()
```

```python
memory = []

def add_message(message):
    memory.append(message)
    if len(memory) > 3:
        memory.pop(0)

def show_memory():
    if memory:
        print("Last 3 messages:")
        for i, message in enumerate(memory, 1):
            print(f"{i}. {message}")
    else:
        print("Memory is empty.")

messages = [
    "I want to watch a movie.",
    "I prefer comedy movies.",
    "I watched 3 Idiots yesterday.",
    "I also like science fiction."
]

for message in messages:
    add_message(message)

show_memory()
```

```python
file_name = "song_history.txt"

def save_song(song):
    with open(file_name, "a", encoding="utf-8") as file:
        file.write(song + "\n")

def load_history():
    try:
        with open(file_name, "r", encoding="utf-8") as file:
            songs = [song.strip() for song in file if song.strip()]
        return songs
    except FileNotFoundError:
        return []

save_song("Blinding Lights - The Weeknd")
save_song("Shape of You - Ed Sheeran")
save_song("Perfect - Ed Sheeran")

history = load_history()

print("Full Play History:")
for i, song in enumerate(history, 1):
    print(f"{i}. {song}")
```

```python
movie_history = [
    {"title": "3 Idiots", "genre": "comedy"},
    {"title": "Dangal", "genre": "drama"},
    {"title": "Hera Pheri", "genre": "comedy"}
]

def get_last_watched():
    return movie_history[-1]

def recommend_similar_movies():
    last_movie = get_last_watched()

    recommendations = {
        "comedy": ["Welcome", "Dhamaal", "Golmaal"],
        "drama": ["Taare Zameen Par", "Swades", "Rockstar"],
        "action": ["War", "Pathaan", "Singham"]
    }

    genre = last_movie["genre"]
    title = last_movie["title"]

    print(f"Last watched movie: {title}")
    print(f"Genre: {genre}")
    print("Recommended similar movies:")

    for movie in recommendations.get(genre, []):
        print("-", movie)

user_query = "comedy movies like the one I watched last week"

if "like" in user_query.lower() and "watched" in user_query.lower():
    recommend_similar_movies()
```
