```python
import requests
import os
from dotenv import load_dotenv

def search_movie(movie_name):
    load_dotenv()
    api_key = os.getenv("OMDB_API_KEY")
    url = "https://www.omdbapi.com/"

    params = {
        "apikey": api_key,
        "t": movie_name
    }

    response = requests.get(url, params=params)
    data = response.json()

    if data.get("Response") == "True":
        print("Title:", data["Title"])
        print("Year:", data["Year"])
        print("IMDb Rating:", data["imdbRating"])
    else:
        print("Movie not found:", data.get("Error"))

movie = input("Enter movie name: ")
search_movie(movie)
```

```python
import pandas as pd
import os

if not os.path.exists("menu.csv"):
    sample_menu = pd.DataFrame({
        "dish": ["Pizza", "Pasta", "Biryani", "Tacos", "Paneer Tikka"],
        "cuisine": ["Italian", "Italian", "Indian", "Mexican", "Indian"],
        "price": [250, 220, 180, 150, 200]
    })
    sample_menu.to_csv("menu.csv", index=False)

menu = pd.read_csv("menu.csv")

cuisine = input("Enter cuisine type: ").strip().lower()

result = menu[menu["cuisine"].str.lower() == cuisine]

if result.empty:
    print("No dishes found.")
else:
    print(f"\n{cuisine.title()} dishes:")
    for _, row in result.iterrows():
        print(f"{row['dish']} - ₹{row['price']}")
```

```python
import sqlite3
import os

connection = sqlite3.connect("music.db")
cursor = connection.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS songs (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT UNIQUE
)
""")

cursor.executemany(
    "INSERT OR IGNORE INTO songs (name) VALUES (?)",
    [("Blinding Lights",), ("Shape of You",), ("Perfect",), ("Believer",)]
)

connection.commit()

song = input("Enter song name: ").strip()

cursor.execute(
    "SELECT name FROM songs WHERE LOWER(name) = LOWER(?)",
    (song,)
)

result = cursor.fetchone()

if result:
    print("Song found")
else:
    print("Song not found")

connection.close()
```

```python
import requests
import os
from dotenv import load_dotenv

load_dotenv()

def get_movies_by_genre(genre):
    api_key = os.getenv("OMDB_API_KEY")
    url = "https://www.omdbapi.com/"

    movies = []
    page = 1

    while len(movies) < 3 and page <= 5:
        params = {
            "apikey": api_key,
            "s": genre,
            "type": "movie",
            "page": page
        }

        data = requests.get(url, params=params).json()

        if data.get("Response") != "True":
            break

        for movie in data.get("Search", []):
            movies.append(movie)

            if len(movies) == 3:
                break

        page += 1

    return movies


def get_movie_details(title):
    api_key = os.getenv("OMDB_API_KEY")

    params = {
        "apikey": api_key,
        "t": title,
        "plot": "full"
    }

    data = requests.get(
        "https://www.omdbapi.com/",
        params=params
    ).json()

    if data.get("Response") == "True":
        print("\nMovie Details")
        print("Title:", data["Title"])
        print("Year:", data["Year"])
        print("Genre:", data["Genre"])
        print("Director:", data["Director"])
        print("Actors:", data["Actors"])
        print("IMDb Rating:", data["imdbRating"])
        print("Plot:", data["Plot"])
    else:
        print("Movie details not found.")


genre = input("Enter movie genre: ")

movies = get_movies_by_genre(genre)

if not movies:
    print("No movies found.")
else:
    print("\nMovies:")
    for i, movie in enumerate(movies, 1):
        print(f"{i}. {movie['Title']} ({movie['Year']})")

    choice = int(input("\nSelect a movie (1-3): "))

    if 1 <= choice <= len(movies):
        selected_movie = movies[choice - 1]["Title"]
        get_movie_details(selected_movie)
    else:
        print("Invalid selection.")
```
