```python
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch

fig, ax = plt.subplots(figsize=(12, 8))
ax.set_xlim(0, 10)
ax.set_ylim(0, 12)
ax.axis("off")

def add_box(x, y, text, width=3, height=1):
    box = FancyBboxPatch(
        (x, y), width, height,
        boxstyle="round,pad=0.08",
        linewidth=1.5
    )
    ax.add_patch(box)
    ax.text(
        x + width / 2, y + height / 2,
        text, ha="center", va="center", fontsize=10
    )

def add_arrow(x1, y1, x2, y2):
    ax.add_patch(
        FancyArrowPatch(
            (x1, y1), (x2, y2),
            arrowstyle="->",
            mutation_scale=15,
            linewidth=1.5
        )
    )

add_box(3.5, 10, "User Request\nCheapest flight\nAhmedabad → Mumbai")
add_box(3.5, 8, "LangChain Agent\nUnderstand Request")
add_box(3.5, 6, "Agent Reasoning\nChoose Flight Search Tool")
add_box(3.5, 4, "Tool Invocation\nFlight Search API")
add_box(3.5, 2, "Compare Prices\nFind Cheapest Flight")
add_box(3.5, 0, "Final Response\nReturn Cheapest Flight")

add_arrow(5, 10, 5, 9)
add_arrow(5, 8, 5, 7)
add_arrow(5, 6, 5, 5)
add_arrow(5, 4, 5, 3)
add_arrow(5, 2, 5, 1)

plt.title("LangChain Agent Workflow: Cheapest Flight Search", fontsize=15)
plt.show()
```

```python
release_years = {
    "Inception": 2010,
    "Avatar": 2009,
    "Titanic": 1997,
    "3 Idiots": 2009
}

trending_songs = {
    2010: "Love the Way You Lie",
    2009: "Poker Face",
    1997: "My Heart Will Go On"
}

def movie_to_song(movie):
    if movie not in release_years:
        print("Movie not found.")
        return

    year = release_years[movie]
    song = trending_songs.get(year, "No song found")

    print("Movie:", movie)
    print("Release Year:", year)
    print("Trending Song:", song)

movie = input("Enter movie name: ")
movie_to_song(movie)
```
