```python
song_search_tool = {
    "name": "search_song",
    "description": "Search for songs by song name and artist.",
    "parameters": {
        "type": "object",
        "properties": {
            "song_name": {
                "type": "string",
                "description": "Name of the song"
            },
            "artist": {
                "type": "string",
                "description": "Name of the artist"
            }
        },
        "required": ["song_name", "artist"]
    }
}

print(song_search_tool)
```

```python
movie_ticket_tool = {
    "name": "find_movie_tickets",
    "description": "Find available movie tickets in a city for a specific movie and date.",
    "parameters": {
        "type": "object",
        "properties": {
            "city": {
                "type": "string",
                "description": "City where the movie will be watched"
            },
            "movie_name": {
                "type": "string",
                "description": "Name of the movie"
            },
            "date": {
                "type": "string",
                "description": "Date of the movie show in YYYY-MM-DD format"
            }
        },
        "required": ["city", "movie_name", "date"]
    }
}

print(movie_ticket_tool)
```

```python
get_weather_tool = {
    "type": "function",
    "function": {
        "name": "GetWeather",
        "description": "Get the weather information for a location and date.",
        "parameters": {
            "type": "object",
            "properties": {
                "location": {
                    "type": "string",
                    "description": "City or location"
                },
                "date": {
                    "type": "string",
                    "description": "Date in YYYY-MM-DD format"
                }
            },
            "required": ["location", "date"],
            "additionalProperties": False
        }
    }
}

print(get_weather_tool)
```

```python
track_order_tool = {
    "name": "trackOrder",
    "description": "Tracks the status of a food delivery order.",
    "parameters": {
        "type": "object",
        "properties": {
            "order_id": {
                "type": "string",
                "description": "Food delivery order ID"
            },
            "user_phone": {
                "type": "string",
                "description": "Phone number associated with the order"
            }
        },
        "required": ["order_id", "user_phone"]
    }
}

print(track_order_tool)
```

```python
The original schema incorrectly used a list for `parameters` and placed each parameter in a separate object. Gemini function calling expects `parameters` to be a schema object containing `type`, `properties`, and `required`. The corrected schema groups `order_id` and `user_phone` inside the `properties` object.
```
