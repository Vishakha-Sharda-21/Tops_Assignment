```python
def food_delivery_agent(user_input):
    text = user_input.lower()

    print("Reasoning: Understanding the user's request and deciding the appropriate action...")

    if "restaurant" in text or "recommend" in text:
        action = "suggest_restaurant"
    elif "menu" in text or "dish" in text:
        action = "show_menu"
    elif "order" in text or "confirm" in text:
        action = "confirm_order"
    else:
        action = "unknown"

    print("External Action:")

    responses = {
        "suggest_restaurant": "I recommend trying a highly rated Italian restaurant nearby.",
        "show_menu": "Here is the restaurant menu with available dishes and prices.",
        "confirm_order": "Your order has been confirmed successfully.",
        "unknown": "Could you please specify whether you want a restaurant recommendation, menu, or order confirmation?"
    }

    print(responses[action])


user_input = input("User: ")
food_delivery_agent(user_input)
```
