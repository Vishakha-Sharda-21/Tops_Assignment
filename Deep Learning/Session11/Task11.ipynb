```python
import tensorflow as tf
from tensorflow.keras.datasets import fashion_mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Flatten, Dense
from tensorflow.keras.optimizers import SGD, RMSprop, Adam
from tensorflow.keras.callbacks import LearningRateScheduler

print("TensorFlow Version:", tf.__version__)
```

```python
# Load dataset
(x_train, y_train), (x_test, y_test) = fashion_mnist.load_data()

# Normalize images
x_train = x_train / 255.0
x_test = x_test / 255.0

print("Training Shape:", x_train.shape)
print("Testing Shape:", x_test.shape)
```

```python
def create_model(optimizer):

    model = Sequential([
        Flatten(input_shape=(28,28)),
        Dense(128, activation='relu'),
        Dense(64, activation='relu'),
        Dense(10, activation='softmax')
    ])

    model.compile(
        optimizer=optimizer,
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )

    return model
```

```python
# Task 1

sgd_model = create_model(SGD())

history_sgd = sgd_model.fit(
    x_train,
    y_train,
    epochs=5,
    validation_data=(x_test, y_test)
)

print("Final Training Accuracy:",
      history_sgd.history['accuracy'][-1])

print("Final Validation Accuracy:",
      history_sgd.history['val_accuracy'][-1])

print("Final Training Loss:",
      history_sgd.history['loss'][-1])

print("Final Validation Loss:",
      history_sgd.history['val_loss'][-1])
```

```python
# Task 2

rms_model = create_model(RMSprop())

history_rms = rms_model.fit(
    x_train,
    y_train,
    epochs=5,
    validation_data=(x_test, y_test)
)

print("SGD Validation Accuracy:",
      history_sgd.history['val_accuracy'][-1])

print("RMSProp Validation Accuracy:",
      history_rms.history['val_accuracy'][-1])
```

```python
# Task 3

adam_optimizer = Adam(learning_rate=0.0005)

adam_model = create_model(adam_optimizer)

history_adam = adam_model.fit(
    x_train,
    y_train,
    epochs=5,
    validation_data=(x_test, y_test)
)

print("Adam Validation Accuracy:",
      history_adam.history['val_accuracy'][-1])
```

```python
# Task 4

initial_lr = 0.01

def scheduler(epoch, lr):

    if epoch != 0 and epoch % 2 == 0:
        return lr * 0.5

    return lr
```

```python
lr_callback = LearningRateScheduler(scheduler)

momentum_optimizer = SGD(
    learning_rate=initial_lr,
    momentum=0.9
)

momentum_model = create_model(momentum_optimizer)

history_momentum = momentum_model.fit(
    x_train,
    y_train,
    epochs=6,
    validation_data=(x_test, y_test),
    callbacks=[lr_callback]
)
```

```python
lr = initial_lr

print("Learning Rate Schedule")

for epoch in range(6):

    if epoch != 0 and epoch % 2 == 0:
        lr = lr * 0.5

    print(f"Epoch {epoch+1}: Learning Rate = {lr}")
```

```python
#Task 5

print("Validation Accuracy Comparison")

print("SGD      :", history_sgd.history['val_accuracy'][-1])
print("RMSProp  :", history_rms.history['val_accuracy'][-1])
print("Adam     :", history_adam.history['val_accuracy'][-1])
```
