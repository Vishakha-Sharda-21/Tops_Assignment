```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.layers import Concatenate
```

```python
model = Sequential([
    Dense(32, activation='relu', input_shape=(10,)),
    Dense(16, activation='relu'),
    Dense(1, activation='sigmoid')
])

model.summary()
```

```python
from tensorflow.keras.layers import Input, Dense
from tensorflow.keras.models import Model

inputs = Input(shape=(10,))

x = Dense(20, activation='relu')(inputs)

outputs = Dense(1)(x)

functional_model = Model(inputs=inputs, outputs=outputs)

functional_model.summary()

print("Total Parameters:",
      functional_model.count_params())
```

```python
age_input = Input(shape=(1,), name="Age")

songs_input = Input(shape=(1,), name="Songs_Listened")

age_branch = Dense(8, activation='relu')(age_input)

songs_branch = Dense(8, activation='relu')(songs_input)

merged = Concatenate()([age_branch, songs_branch])

hidden = Dense(16, activation='relu')(merged)

output = Dense(
    1,
    activation='sigmoid',
    name="Premium_Prediction"
)(hidden)
```

```python
multi_input_model = Model(
    inputs=[age_input, songs_input],
    outputs=output
)

multi_input_model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

multi_input_model.summary()
```
