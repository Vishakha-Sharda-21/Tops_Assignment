```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# Create Sequential Model
model = Sequential([
    Dense(16, activation='relu', input_shape=(2,)),
    Dense(8, activation='relu'),
    Dense(1, activation='sigmoid')
])

# Display Model Summary
model.summary()
```

```python
# Compile Model

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

print("Model compiled successfully!")
```

```python
import numpy as np

# For reproducibility
np.random.seed(42)

# Generate 100 random user profiles

# Feature 1: Listening Time (0–100 minutes)
listening_time = np.random.rand(100) * 100

# Feature 2: Genre Score (0–1)
liked_genre_score = np.random.rand(100)

# Combine Features
X = np.column_stack((listening_time, liked_genre_score))

# Generate Binary Labels
y = np.random.randint(0, 2, size=100)

# Train Model
history = model.fit(
    X,
    y,
    epochs=20,
    validation_split=0.2,
    verbose=1
)

# Print Final Accuracy
print("\nFinal Training Accuracy:",
      history.history['accuracy'][-1])
```

```python
from tensorflow.keras.callbacks import EarlyStopping

# Early Stopping Callback
early_stop = EarlyStopping(
    monitor='val_loss',
    patience=3,
    restore_best_weights=True
)

# Train Again Using Callback
history = model.fit(
    X,
    y,
    epochs=20,
    validation_split=0.2,
    callbacks=[early_stop],
    verbose=1
)

# Print Final Accuracy
print("\nFinal Training Accuracy:",
      history.history['accuracy'][-1])

print("Training completed with EarlyStopping.")
```
