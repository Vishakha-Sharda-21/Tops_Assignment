```python
import tensorflow as tf
from tensorflow.keras.layers import Input, Dense, Concatenate, Add
from tensorflow.keras.models import Model

# Input Layers
user_age = Input(shape=(1,), name="User_Age")
favorite_genre = Input(shape=(1,), name="Favorite_Genre")

# Combine Inputs
combined = Concatenate()([user_age, favorite_genre])

# Hidden Layers
x = Dense(16, activation='relu')(combined)
x = Dense(8, activation='relu')(x)

# Output Layer (Predicted Rating between 1 and 5)
movie_rating = Dense(1, activation='linear', name="Movie_Rating")(x)

# Create Model
movie_model = Model(
    inputs=[user_age, favorite_genre],
    outputs=movie_rating
)

# Compile Model
movie_model.compile(
    optimizer='adam',
    loss='mse',
    metrics=['mae']
)

# Display Summary
movie_model.summary()
```

```python
# Inputs
order_amount = Input(shape=(1,), name="Order_Amount")
distance = Input(shape=(1,), name="Distance")

# Combine Inputs
combined = Concatenate()([order_amount, distance])

# Shared Hidden Layers
x = Dense(32, activation='relu')(combined)
x = Dense(16, activation='relu')(x)

# Output 1: Delivery Time
delivery_time = Dense(
    1,
    activation='linear',
    name="Delivery_Time"
)(x)

# Output 2: Delivery Rating
delivery_rating = Dense(
    1,
    activation='linear',
    name="Delivery_Rating"
)(x)

# Create Model
zomato_model = Model(
    inputs=[order_amount, distance],
    outputs=[delivery_time, delivery_rating]
)

# Compile
zomato_model.compile(
    optimizer='adam',
    loss={
        "Delivery_Time": "mse",
        "Delivery_Rating": "mse"
    },
    metrics={
        "Delivery_Time": ["mae"],
        "Delivery_Rating": ["mae"]
    }
)

# Summary
zomato_model.summary()
```

```python

playlist_input = Input(shape=(10,), name="Playlist_Features")

# Hidden Layers
x = Dense(16, activation='relu')(playlist_input)
x = Dense(8, activation='relu')(x)

# Output Layer
playlist_output = Dense(
    1,
    activation='sigmoid',
    name="Recommendation"
)(x)

# Functional API Model
playlist_model = Model(
    inputs=playlist_input,
    outputs=playlist_output
)

# Compile
playlist_model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# Summary
playlist_model.summary()
```

```python
# Input Layer
ipl_input = Input(shape=(5,), name="IPL_Input")

# Hidden Layer
hidden = Dense(16, activation='relu')(ipl_input)

# Intermediate Output
hidden_output = Dense(1, activation='linear')(hidden)

# Direct Connection from Input
direct_output = Dense(1, activation='linear')(ipl_input)

# Skip Connection
final_output = Add(name="Predicted_Score")(
    [hidden_output, direct_output]
)

# Create Model
ipl_model = Model(
    inputs=ipl_input,
    outputs=final_output
)

# Compile
ipl_model.compile(
    optimizer='adam',
    loss='mse',
    metrics=['mae']
)

# Display Summary
ipl_model.summary()
```
