```python
import numpy as np

# 5x5 Grayscale Image
image = np.array([
    [1, 2, 3, 0, 1],
    [4, 5, 6, 1, 2],
    [7, 8, 9, 2, 3],
    [1, 2, 3, 4, 5],
    [6, 7, 8, 9, 0]
])

# 3x3 Filter
kernel = np.array([
    [1, 0, -1],
    [1, 0, -1],
    [1, 0, -1]
])

# Output Size
output = np.zeros((3,3))

# Convolution Operation
for i in range(output.shape[0]):
    for j in range(output.shape[1]):
        region = image[i:i+3, j:j+3]
        output[i,j] = np.sum(region * kernel)

print("Input Image:\n", image)

print("\nKernel:\n", kernel)

print("\nOutput Feature Map:\n", output)
```

```python
import numpy as np

def max_pooling(image, pool_size=2):

    rows, cols = image.shape

    output_rows = rows // pool_size
    output_cols = cols // pool_size

    pooled = np.zeros((output_rows, output_cols))

    for i in range(output_rows):
        for j in range(output_cols):

            region = image[
                i*pool_size:(i+1)*pool_size,
                j*pool_size:(j+1)*pool_size
            ]

            pooled[i, j] = np.max(region)

    return pooled

# Example Image
image = np.array([
    [1,3,2,1],
    [4,6,5,2],
    [7,8,9,3],
    [2,4,1,5]
])

print("Original Image:\n", image)

pooled = max_pooling(image)

print("\nMax Pooled Image:\n", pooled)
```
