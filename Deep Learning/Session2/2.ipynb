```python
import numpy as np 

def perceptron(inputs, weights, bias): 
    z = np.dot(inputs, weights) + bias 
    output = 1 if z >= 0 else 0 
    
    return output 

weights = np.array([0.5, 0.3])
```

```python
weights = np.array([0.5, 0.3]) 

test_cases = [ (np.array([2,1]), -1), (np.array([0,1]), -1), (np.array([4,2]), -1), ]
```

```python
for inputs, bias in test_cases:
    result = perceptron(inputs, weights, bias) 
print("Input:", inputs, "Bias:", bias, "Output:", result)
```
