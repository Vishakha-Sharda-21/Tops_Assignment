```python
import numpy as np 

X = np.array([ [0,0], [0,1], [1,0], [1,1] ]) 
y = np.array([0,1,1,0])
```

```python
weights = np.zeros(2) 
bias = 0 
lr = 0.1
```

```python
def step(x):
    return 1 if x >= 0 else 0

for epoch in range(100):

    for i in range(len(X)):

        pred = step(np.dot(X[i], weights) + bias)

        error = y[i] - pred

        weights += lr * error * X[i]

        bias += lr * error

print("Weights:", weights)
print("Bias:", bias)

for x in X:
    print(x, step(np.dot(x, weights) + bias))
```
