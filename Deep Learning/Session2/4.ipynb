```python
import numpy as np 

def sigmoid(x): 
    return 1/(1+np.exp(-x)) 

X = np.array([ [0,0], [0,1], [1,0], [1,1] ]) 
W1 = np.array([ [20,-20], [20,-20] ])
```

```python
b1 = np.array([-10,30]) 
W2 = np.array([ [20], [20] ]) 
b2 = np.array([-30])
```

```python
for x in X:
    hidden = sigmoid(np.dot(x,W1)+b1) 
    output = sigmoid(np.dot(hidden,W2)+b2) 
    prediction = round(float(output)) 
print(x, prediction)
```
