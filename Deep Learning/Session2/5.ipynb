```python
import numpy as np 
def sigmoid(x): 
    return 1/(1+np.exp(-x)) 


X = np.array([ [4.8,500], [4.6,420], [3.2,50], [3.5,80] ])
W1 = np.array([ [0.4,0.2], [0.01,0.03] ])
```

```python
b1 = np.array([0.1,0.2]) 
W2 = np.array([ [0.5], [0.7] ]) 
b2 = np.array([0.1])
```

```python
for sample in X: 
    hidden = sigmoid(np.dot(sample,W1)+b1) 
    output = sigmoid(np.dot(hidden,W2)+b2) 
    prediction = 1 if output >= 0.5 else 0 
    print("Input:", sample) 
    print("Output Probability:", output[0]) 
    print("Prediction:", prediction) 
    print()
```
