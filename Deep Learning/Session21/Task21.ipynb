```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D

# Create CNN Model
model = Sequential()

# Add Conv2D Layer
model.add(
    Conv2D(
        filters=32,
        kernel_size=(3, 3),
        activation='relu',
        input_shape=(32, 32, 3)
    )
)

# Display Model Summary
model.summary()
```

```python
from tensorflow.keras.layers import MaxPooling2D, Flatten, Dense

# Add MaxPooling Layer
model.add(MaxPooling2D(pool_size=(2, 2)))

# Flatten Layer
model.add(Flatten())

# Dense Layer
model.add(Dense(64, activation='relu'))

# Output Layer (10 Classes)
model.add(Dense(10, activation='softmax'))

# Display Summary
model.summary()
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Conv2D,
    BatchNormalization,
    MaxPooling2D,
    Flatten,
    Dense,
    Dropout
)

# Create New Model
model = Sequential([
    Conv2D(
        32,
        (3, 3),
        activation='relu',
        input_shape=(32, 32, 3)
    ),

    # Batch Normalization
    BatchNormalization(),

    # Max Pooling
    MaxPooling2D((2, 2)),

    # Flatten
    Flatten(),

    # Dropout
    Dropout(0.3),

    # Dense Layer
    Dense(64, activation='relu'),

    # Output Layer
    Dense(10, activation='softmax')
])

# Display Summary
model.summary()
```

```python
import numpy as np
from tensorflow.keras.utils import to_categorical

# Compile Model
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# Generate Random Images
X_train = np.random.rand(500, 32, 32, 3).astype(np.float32)

# Generate Random Labels (0-9)
y_train = np.random.randint(0, 10, 500)

# One-Hot Encode Labels
y_train = to_categorical(y_train, num_classes=10)

# Train Model
history = model.fit(
    X_train,
    y_train,
    epochs=5,
    batch_size=32,
    validation_split=0.2
)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Conv2D,
    BatchNormalization,
    MaxPooling2D,
    Flatten,
    Dense,
    Dropout
)

# Improved CNN Model
improved_model = Sequential([

    Conv2D(
        32,
        (3,3),
        activation='relu',
        input_shape=(32,32,3)
    ),
    BatchNormalization(),
    MaxPooling2D((2,2)),

    # Additional Convolution Block
    Conv2D(
        64,
        (3,3),
        activation='relu'
    ),
    BatchNormalization(),
    MaxPooling2D((2,2)),

    Flatten(),

    Dropout(0.3),

    Dense(128, activation='relu'),

    Dense(10, activation='softmax')
])

# Compile
improved_model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# Display Summary
improved_model.summary()
```
