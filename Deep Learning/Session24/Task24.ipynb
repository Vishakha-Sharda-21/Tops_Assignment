```python
import numpy as np

# Sequence (Spotify song play counts)
sequence = [10, 15, 12, 18, 20]

# Initialize parameters
W = 0.5
U = 0.8
b = 0.1

# Initial hidden state
h = 0

print("Processing Sequence:\n")

for t, x in enumerate(sequence):
    h = np.tanh(W * x + U * h + b)
    print(f"Step {t+1}: Input = {x}, Hidden State = {h:.4f}")

print("\nFinal Hidden State:", h)
```

```python
import numpy as np

sentence = "I love cricket and IPL matches"

# Split sentence into words
words = sentence.split()

# Create word-to-index mapping
word_to_index = {word: idx + 1 for idx, word in enumerate(words)}

print("Word to Index Mapping:")
print(word_to_index)

# Convert sentence to integer sequence
sequence = [word_to_index[word] for word in words]

print("\nInteger Sequence:")
print(sequence)

# RNN Parameters
W = 0.4
U = 0.7
b = 0.2

# Initial Hidden State
h = 0

print("\nHidden States:")

for index in sequence:
    h = np.tanh(W * index + U * h + b)
    print(f"Input = {index}, Hidden State = {h:.4f}")
```

```python
print("Vanishing Gradient Problem\n")

print("The vanishing gradient problem occurs when gradients become extremely")
print("small during training. As a result, the early layers or earlier time")
print("steps of an RNN learn very slowly or stop learning altogether.")

print("\nReal-world Example:")
print("In speech recognition, an RNN may forget words spoken earlier in a")
print("long sentence, making it difficult to correctly recognize the complete")
print("speech or understand the overall context.")
```

```python
import numpy as np

# Simple 3-step sequence
inputs = [1, 2, 3]

# Parameters
W = 0.5
U = 0.8
b = 0.1

# Forward Pass
hidden_states = [0]

for x in inputs:
    h = np.tanh(W * x + U * hidden_states[-1] + b)
    hidden_states.append(h)

print("Hidden States:")
print(hidden_states)

# Assume gradient from output
grad = 1.0

print("\nBackward Pass (BPTT):")

for t in reversed(range(1, len(hidden_states))):
    grad = grad * (1 - hidden_states[t]**2) * U
    print(f"Gradient at Time Step {t}: {grad:.4f}")
```
