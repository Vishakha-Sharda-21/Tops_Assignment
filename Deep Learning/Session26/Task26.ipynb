```python
import numpy as np

# Fake Flipkart product prices for 10 days
prices = np.array([520, 525, 530, 528, 535, 540, 545, 550, 548, 555])

# Create sequences
sequence_length = 3

X = []
y = []

for i in range(len(prices) - sequence_length):
    X.append(prices[i:i + sequence_length])
    y.append(prices[i + sequence_length])

X = np.array(X)
y = np.array(y)

# Reshape for LSTM (samples, time steps, features)
X = X.reshape((X.shape[0], X.shape[1], 1))

print("Input Shape:", X.shape)
print("Target Shape:", y.shape)

print("\nSample Input:")
print(X[0])

print("Sample Target:", y[0])
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense

# Build model
lstm_model = Sequential([
    LSTM(32, input_shape=(3,1)),
    Dense(1)
])

lstm_model.compile(
    optimizer='adam',
    loss='mse'
)

# Train model
history = lstm_model.fit(
    X,
    y,
    epochs=10,
    verbose=1
)
```

```python
# Predict using the last sequence
test_sequence = X[-1].reshape(1,3,1)

predicted_price = lstm_model.predict(test_sequence)

print("Actual Price :", y[-1])
print("Predicted Price :", predicted_price[0][0])
```

```python
import numpy as np

def prepare_sequence(data, sequence_length):
    """
    Convert a sequence into LSTM input format.
    """

    X = []
    y = []

    for i in range(len(data) - sequence_length):
        X.append(data[i:i+sequence_length])
        y.append(data[i+sequence_length])

    X = np.array(X)
    y = np.array(y)

    X = X.reshape((X.shape[0], X.shape[1], 1))

    return X, y


# Example: Spotify daily song play counts
song_counts = [20,25,30,28,35,40,38,45,50]

X, y = prepare_sequence(song_counts,3)

print("Input Shape:", X.shape)
print("Target Shape:", y.shape)
```
