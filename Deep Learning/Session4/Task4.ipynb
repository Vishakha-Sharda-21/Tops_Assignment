```python
# Task 1

import numpy as np

# Sigmoid Function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Tanh Function
def tanh(x):
    return np.tanh(x)

# ReLU Function
def relu(x):
    return max(0, x)

# Testing

x = 2.5

print("Input:", x)
print("Sigmoid:", sigmoid(x))
print("Tanh:", tanh(x))
print("ReLU:", relu(x))
```

```python
# Task 2

import matplotlib.pyplot as plt

# Activation Functions

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def tanh(x):
    return np.tanh(x)

def relu(x):
    return np.maximum(0, x)

# Input Values

x = np.linspace(-10, 10, 1000)

# Plot

plt.figure(figsize=(8,5))

plt.plot(x, sigmoid(x), label="Sigmoid")
plt.plot(x, tanh(x), label="Tanh")
plt.plot(x, relu(x), label="ReLU")

plt.title("Activation Functions")
plt.xlabel("Input")
plt.ylabel("Output")

plt.legend()
plt.grid(True)

plt.show()
```

```python
# Task 3


# Sigmoid

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Derivative of Sigmoid

def sigmoid_derivative(x):
    s = sigmoid(x)
    return s * (1 - s)

# Values from -10 to 10

inputs = np.arange(-10, 11)

print("Input\tDerivative")

for x in inputs:
    print(f"{x}\t{sigmoid_derivative(x):.6f}")
```

```python
# Task 4

# ReLU Function

def relu(x):
    return np.maximum(0, x)

# Negative Inputs

inputs = np.array([
    -10,
    -8,
    -5,
    -3,
    -1
])

outputs = relu(inputs)

print("Input Array:")
print(inputs)

print("\nOutput Array:")
print(outputs)
```

```python
# Task 5

# Original Sigmoid

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Modified Sigmoid

def scaled_sigmoid(x, a):
    return 1 / (1 + np.exp(-(a * x)))

# Input Values

x = np.linspace(-10, 10, 1000)

# Outputs

original = sigmoid(x)

scaled = scaled_sigmoid(x, 2)

# Plot

plt.figure(figsize=(8,5))

plt.plot(
    x,
    original,
    label="Original Sigmoid"
)

plt.plot(
    x,
    scaled,
    label="Scaled Sigmoid (a=2)"
)

plt.title("Original vs Scaled Sigmoid")
plt.xlabel("Input")
plt.ylabel("Output")

plt.legend()
plt.grid(True)

plt.show()
```
