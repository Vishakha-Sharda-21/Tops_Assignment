```python
# Task 1

import numpy as np

def leaky_relu(x, alpha=0.01):
    return np.where(x > 0, x, alpha * x)

# Test values from -5 to 5

inputs = np.arange(-5, 6)

outputs = leaky_relu(inputs)

print("Input Values:")
print(inputs)

print("\nLeaky ReLU Outputs:")
print(outputs)
```

```python
# Task 2

import matplotlib.pyplot as plt

# ReLU

def relu(x):
    return np.maximum(0, x)

# Leaky ReLU

def leaky_relu(x, alpha=0.01):
    return np.where(x > 0, x, alpha * x)

# Parametric ReLU

def parametric_relu(x, a):
    return np.where(x > 0, x, a * x)

# Input Range

x = np.linspace(-5, 5, 1000)

# Outputs

relu_output = relu(x)
leaky_output = leaky_relu(x)
prelu_output = parametric_relu(x, 0.2)

# Plot

plt.figure(figsize=(8,5))

plt.plot(x, relu_output, label="ReLU")
plt.plot(x, leaky_output, label="Leaky ReLU (α=0.01)")
plt.plot(x, prelu_output, label="PReLU (a=0.2)")

plt.title("ReLU vs Leaky ReLU vs PReLU")
plt.xlabel("Input")
plt.ylabel("Output")

plt.grid(True)
plt.legend()

plt.show()
```

```python
# Task 3

def softmax(scores):

    exp_scores = np.exp(scores)

    probabilities = exp_scores / np.sum(exp_scores)

    return probabilities

scores = np.array([2.0, 1.0, 0.1, 0.5])

probs = softmax(scores)

print("Softmax Probabilities:")

for p in probs:
    print(round(p, 3))
```

```python
# Task 4

def hard_sigmoid(x):

    return max(
        0,
        min(
            1,
            0.2 * x + 0.5
        )
    )

inputs = [-3, -1, 0, 1, 3]

for x in inputs:
    print(
        f"Input: {x} -> Output: {hard_sigmoid(x)}"
    )
```
