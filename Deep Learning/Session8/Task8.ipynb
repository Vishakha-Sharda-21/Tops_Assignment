```python
# Task 1

def compute_gradient(f, x):

    h = 1e-5

    gradient = (f(x + h) - f(x - h)) / (2 * h)

    return gradient


# Example Function

def f(x):
    return x ** 2


x = 3

grad = compute_gradient(f, x)

print("Function: f(x) = x^2")
print("x =", x)
print("Gradient =", grad)
```

```python
# Task 2

def partial_x(x, y):
    return 2 * x

def partial_y(x, y):
    return 2 * y


x = 3
y = 4

dx = partial_x(x, y)
dy = partial_y(x, y)

print("Function: f(x,y) = x^2 + y^2")
print("Point: (3,4)")
print("∂f/∂x =", dx)
print("∂f/∂y =", dy)
```

```python
# Task 3


learning_rate = 0.1

w = 0

gradient = 2 * (w - 5)

w_new = w - learning_rate * gradient

print("Initial w =", w)
print("Gradient =", gradient)
print("Updated w =", w_new)
```

```python
# Task 4

import matplotlib.pyplot as plt

learning_rates = [0.01, 0.1, 0.5]

iterations = 20

for lr in learning_rates:

    x = 0

    history = []

    for i in range(iterations):

        history.append(x)

        gradient = 2 * (x - 3)

        x = x - lr * gradient

    plt.plot(
        range(iterations),
        history,
        label=f"LR={lr}"
    )

plt.title("Gradient Descent Convergence")
plt.xlabel("Iteration")
plt.ylabel("x Value")

plt.legend()

plt.grid(True)

plt.show()
```

```python
# Task 5

import numpy as np

def f(x):
    return x**4 - 3*x**3 + 2

# Generate x values

x = np.linspace(-2, 4, 1000)

y = f(x)

plt.figure(figsize=(8,5))

plt.plot(x, y)

plt.title("f(x) = x^4 - 3x^3 + 2")

plt.xlabel("x")
plt.ylabel("f(x)")

plt.grid(True)

# Mark Approximate Global Minimum

global_x = 2.25
global_y = f(global_x)

plt.scatter(
    global_x,
    global_y,
    label="Global Minimum"
)

plt.legend()

plt.show()

print("Approximate Global Minimum:")
print("x =", global_x)
print("f(x) =", global_y)
```
