```python
# Task 2

import numpy as np

# Dummy Data

X = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

y = np.array([0,1,1,1])

# Model

model = Sequential([
    Dense(2, activation='relu', input_shape=(2,)),
    Dense(1, activation='sigmoid')
])

model.compile(
    optimizer='sgd',
    loss='binary_crossentropy'
)

# Build model

model.predict(X, verbose=0)

print("Weights BEFORE training:")
for layer in model.layers:
    print(layer.get_weights())

# Train one epoch

model.fit(
    X,
    y,
    epochs=1,
    verbose=0
)

print("\nWeights AFTER training:")
for layer in model.layers:
    print(layer.get_weights())
```

```python
# Task 3

import math

# Given Values

x = 2
w = 0.5
b = 1
y = 1

# Forward Pass

z = x*w + b

a = 1/(1+math.exp(-z))

# Loss = (a-y)^2

loss = (a-y)**2

# Chain Rule

dL_da = 2*(a-y)

da_dz = a*(1-a)

dz_dw = x

dL_dw = dL_da * da_dz * dz_dw

print("z =", z)
print("a =", a)
print("Loss =", loss)

print("\ndL/da =", dL_da)
print("da/dz =", da_dz)
print("dz/dw =", dz_dw)

print("\ndL/dw =", dL_dw)
```

```python
# Task 4

import math

def sigmoid(x):
    return 1/(1+math.exp(-x))

x = 0.9

print("Initial Value:", x)

for layer in range(10):

    x = sigmoid(x)

    print(
        f"Layer {layer+1}: {x}"
    )

print("\nFinal Output:", x)
```

```python
# Task 5

import numpy as np

# Large Initial Weights

weights = np.array([
    [10,10],
    [10,10]
])

# Input

x = np.array([
    1,
    1
])

# Simulate Deep Layers

activation = x

for layer in range(5):

    activation = np.dot(
        activation,
        weights
    )

    print(
        f"Layer {layer+1}:",
        activation
    )
```
