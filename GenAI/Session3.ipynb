```python
# Task 3

import numpy as np

def simple_generator(random_vector):
    return np.random.rand(28, 28)

class Discriminator:

    def discriminate(self, image):
        mean_pixel = np.mean(image)

        if mean_pixel < 0.5:
            return 0.9
        else:
            return 0.1

random_vector = np.random.randn(100)
fake_image = simple_generator(random_vector)

discriminator = Discriminator()
score = discriminator.discriminate(fake_image)

print("Discriminator score:", score)

if score < 0.5:
    print("Discriminator correctly identified the image as FAKE.")
else:
    print("Discriminator incorrectly identified the image as REAL.")
```
