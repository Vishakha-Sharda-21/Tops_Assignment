```python
# Task 1

import cv2
import numpy as np
import os

def forward_diffusion(image_path):
    image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

    if image is None:
        raise ValueError("Image not found")

    image = image.astype(np.float32) / 255.0

    os.makedirs("diffusion_steps", exist_ok=True)

    for step in range(1, 11):
        noise = np.random.normal(
            0,
            0.05 * step,
            image.shape
        )

        noisy_image = image + noise
        noisy_image = np.clip(noisy_image, 0, 1)

        output = (noisy_image * 255).astype(np.uint8)

        cv2.imwrite(
            f"diffusion_steps/step_{step}.png",
            output
        )

    print("10 noisy images saved successfully.")


forward_diffusion("input.jpg")
```

```python
# Task 2

import cv2

original = cv2.imread(
    "input.jpg",
    cv2.IMREAD_GRAYSCALE
)

noisy = cv2.imread(
    "diffusion_steps/step_10.png",
    cv2.IMREAD_GRAYSCALE
)

denoised = cv2.GaussianBlur(
    noisy,
    (5, 5),
    0
)

cv2.imwrite(
    "denoised.png",
    denoised
)

cv2.imshow("Original", original)
cv2.imshow("Noisy", noisy)
cv2.imshow("Denoised", denoised)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
