```python
# Task 2

def restaurant_chatbot(name, cuisine):
    restaurants = {
        "Pizza Palace": "Italian",
        "Spice Garden": "Indian",
        "Sushi World": "Japanese",
        "Burger Hub": "American"
    }

    if name in restaurants:
        if restaurants[name].lower() == cuisine.lower():
            return f"Yes, {name} serves {cuisine} cuisine."
        else:
            return f"No, {name} serves {restaurants[name]} cuisine."

    return f"Sorry, I don't have information about {name}."


name = input("Enter restaurant name: ")
cuisine = input("Enter cuisine: ")

print(restaurant_chatbot(name, cuisine))
```

```python
# Task 4

def filter_songs(songs):
    return [song for song in songs if song.startswith("S")]

songs = [
    "Shape of You",
    "Perfect",
    "Senorita",
    "Believer",
    "Stay"
]

result = filter_songs(songs)

print(result)
```
